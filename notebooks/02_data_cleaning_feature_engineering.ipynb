```python
import pandas as pd
import numpy as np

pd.set_option("display.max_columns", None)

df = pd.read_csv("../data/bank_customer_churn.csv")
df.head()
```

```python
# rename columns
df = df.rename(columns={
    "Satisfaction Score": "SatisfactionScore",
    "Card Type": "CardType",
    "Point Earned": "PointEarned"
})
```

```python
# drop unnecessary columns
df_clean = df.drop(columns=["RowNumber", "CustomerId", "Surname"])
```

```python
# create age groups
df_clean["AgeGroup"] = pd.cut(
    df_clean["Age"],
    bins=[17, 30, 40, 50, 60, 100],
    labels=["18-30", "31-40", "41-50", "51-60", "60+"]
)
```

```python
# create balance groups
df_clean["BalanceGroup"] = pd.cut(
    df_clean["Balance"],
    bins=[-1, 0, 50000, 100000, 150000, 300000],
    labels=["No Balance", "Low", "Medium", "High", "Very High"]
)
```

```python
# create tenure groups
df_clean["TenureGroup"] = pd.cut(
    df_clean["Tenure"],
    bins=[-1, 2, 5, 8, 10],
    labels=["New", "Medium", "Long", "Very Long"]
)
```

```python
# create salary groups
df_clean["SalaryGroup"] = pd.cut(
    df_clean["EstimatedSalary"],
    bins=[0, 50000, 100000, 150000, 200000],
    labels=["Low", "Medium", "High", "Very High"]
)
```

```python
# save cleaned dataset
df_clean.to_csv("../data/cleaned_bank_churn.csv", index=False)
```
