## Customer Churn Distribution

This chart shows the overall split between retained and churned customers. It helps establish the baseline churn problem before analysing specific customer segments.

```python
# quick target visualisation
ax = sns.countplot(
    data=df,
    x="Exited",
    hue="Exited",
    palette=["#4CAF50", "#E74C3C"],
    legend=False
)

# add count labels
for p in ax.patches:
    ax.annotate(
        f"{int(p.get_height()):,}",
        (p.get_x() + p.get_width() / 2, p.get_height()),
        ha="center",
        va="bottom",
        fontsize=10,
    )

plt.title("Customer Churn Distribution")
plt.xlabel("Customer Status")
plt.ylabel("Number of Customers")
plt.xticks([0, 1], ["Retained", "Churned"])
plt.show()
```

### Insight

Most customers are retained, but around one-fifth of customers have churned. This shows that the dataset is moderately imbalanced and churn prediction should focus on identifying the smaller churned group accurately.
